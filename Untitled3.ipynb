```python
import re, unicodedata
from nltk.chat.util import Chat

SOLICITUDES = {
    "SOL-001": {"estatus": "SOLICITADO", "detalle": "Pendiente de asignación"},
    "SOL-002": {"estatus": "PROGRAMADO", "detalle": "Proveedor y unidad asignados"},
    "SOL-003": {"estatus": "EN RUTA", "detalle": "Transporte en movimiento"},
    "SOL-004": {"estatus": "ENTREGADO", "detalle": "Servicio concluido"}
}

def normalizar(texto):
    texto = unicodedata.normalize("NFD", texto.lower().strip())
    texto = "".join(c for c in texto if unicodedata.category(c) != "Mn")
    return re.sub(r"\s+", " ", texto)

def consultar_solicitud(texto):
    m = re.search(r"\bSOL-\d{3}\b", texto.upper())
    if not m:
        return "Indica el folio con formato SOL-001."
    folio = m.group(0)
    dato = SOLICITUDES.get(folio)
    if not dato:
        return f"{folio} no existe en la base de prueba."
    return f"{folio}: {dato['estatus']}. {dato['detalle']}."

# Todos los patrones deben estar en minusculas y sin acentos para coincidir con la normalizacion
pares = [
 (r"^(hola|buen dia|buenas tardes|buenos dias|buenas noches|que tal|ey).*$",
  ["Hola, soy ATMA. Puedo ayudarte con acceso, nuevas solicitudes, "
   "disponibilidad y estatus."]),
 (r".*(ayuda|opciones|que puedes hacer|menu).*",
  ["Opciones: acceso, nueva solicitud, datos requeridos, "
   "disponibilidad, estatus y contacto con logistica."]),
 (r".*(nueva solicitud|solicitar transporte|necesito transporte|pedir transporte|crear solicitud|nuevas solicitudes).*",
  ["Registra la necesidad en la app Solicitud de Transporte de "
   "Maquinaria. Prepara fecha, equipo, origen, destino y responsables."]),
 (r".*(donde|plataforma|power apps|app|aplicacion).*(solicitud|registrar|ingresar|entrar).*",
  ["Las solicitudes se registran en Power Apps, en la aplicacion "
   "Solicitud de Transporte de Maquinaria."]),
 (r".*(acceso|no tengo acceso|sin acceso|no puedo entrar|permiso|error al entrar|bloqueado).*",
  ["Identifica si no ves la app, falta permiso o aparece un error. "
   "El acceso debe escalarse al responsable de Power Apps."]),
 (r".*(datos|informacion|capturar|requisitos|que necesito|origen|destino).*",
  ["Captura fecha y hora, equipo y numero economico, origen, destino, "
   "horario y responsables de carga y descarga."]),
 (r".*(disponibilidad|hay transporte|unidad disponible|cama baja|unidades).*",
  ["Para revisar disponibilidad indica fecha y hora, equipo, origen, "
   "destino y tipo de unidad. La confirmacion final corresponde a logistica."]),
 (r".*(foraneo|fuera de la zona|foranea).*",
  ["Los movimientos foraneos deben solicitarse con al menos tres dias "
   "habiles de anticipacion por la gestion de permisos."]),
 (r".*(urgente|cambiar|cancelar|cancelacion|modificar|logistica|contacto).*",
  ["El caso requiere intervencion humana. Comparte el folio y el motivo "
   "con el equipo de logistica."]),
 (r"^(gracias|entendido|perfecto|excelente|gracias de verdad).*$", ["Con gusto."]),
 (r"^(salir|adios|fin|bye).*$", ["Conversacion finalizada."]),
 (r".*", ["No identifique la consulta. Escribe: acceso, nueva solicitud, "
            "disponibilidad, estatus o logistica."])
]

bot = Chat(pares, reflections={})

def responder(mensaje):
    limpio = normalizar(mensaje)
    # Detectar si el usuario incluyo un folio como SOL-123 directamente o pregunto por estatus
    if re.search(r"\bsol-\d{3}\b", limpio) or re.search(r"\b(estatus|estado|seguimiento|folio)\b", limpio):
        return consultar_solicitud(mensaje)
    return bot.respond(limpio)

def iniciar_chat():
    print("ATMA - Asistente de Transporte de Maquinaria (Ejecutando)")
    while True:
        mensaje = input("Usuario: ").strip()
        if not mensaje:
            print("ATMA: Escribe una consulta.")
            continue
        respuesta = responder(mensaje)
        print("ATMA:", respuesta)
        if normalizar(mensaje) in {"salir", "adios", "fin"}:
            break

iniciar_chat()
```
